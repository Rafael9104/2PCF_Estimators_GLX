```python
#Estableciendo la lista de números aleatorios los cuales servirán de coordenadas para establecer los puntos aleatorios en 
#el intervalo entre (-500,500), para cada configuración requerida: disco, anillo y anillos
x_list1 = np.random.uniform(-500,500,60000)
y_list1 = np.random.uniform(-500,500,60000)
```

```python
#zip comprime las listas en un solo elemento
radius_disc = 300
counter = 0
x_disc = []
y_disc = []
for i,j in zip(x_list1,y_list1):
    if np.sqrt(i*i+j*j) < radius_disc:
        if counter < 1000:
            counter += 1
            x_disc.append(i)
            y_disc.append(j)
x_disc = np.array(x_disc)
y_disc = np.array(y_disc)
print(f"El número de puntos dentro del disco son: {len(x_disc)}")
# con este comando acomodamos los datos que se guardaran en el archivo .dat o ,txt, de tal forma que se guarden en columnas.
dataout_disc = np.column_stack((x_disc, y_disc)) 
# con delimiter aplicamos un espaciado entre columnas de datos en el archivo que se guardará.
np.savetxt(path + names[1] + '_data.npy', dataout_disc, delimiter = '\t', header = header_x + header_y)
```

4. d) 50 anillos con 100 puntos cada uno cuyos centros están distribuidos aleatoriamente sobre la caja

```python
radius_rings = 150
```

```python
counter = 0
x_centers = []
y_centers = []
for i,j in zip(x_list1, y_list1):
    if 500 - radius_rings > i > -500 + radius_rings:
        if 500 - radius_rings > j > -500 + radius_rings:
            if 50 > counter:
                counter += 1
                x_centers.append(i)
                y_centers.append(j)
x_centers = np.array(x_centers)
y_centers = np.array(y_centers)
print(f"El número de centros para los anillos es: {len(x_centers)}")
```

```python
x_rings = []
y_rings = []
for i,j in zip(x_centers, y_centers):
    counter = 0
    x_1 = np.random.uniform(i - radius_rings, i + radius_rings, 10000)
    y_1 = np.random.uniform(j - radius_rings, j + radius_rings, 10000)
    for k,l in zip(x_1, y_1):
        x_2 = k - i
        y_2 = l - j
        if radius_rings > np.sqrt(x_2*x_2 + y_2*y_2) > radius_rings - delta_radius:
            if 100 > counter:
                counter += 1
                x_rings.append(k)
                y_rings.append(l)
x_rings = np.array(x_rings)
y_rings = np.array(y_rings)
```

### Disc ($N_R = N_D$)

```python
plt.figure(figsize=(10,10))
font1={'size':16}
font2={'family':'serif','size':20}
plt.xlabel('$X$', fontdict=font1)
plt.ylabel('$Y$', fontdict=font1)
plt.title('Points D and R of disk', fontdict=font2)
plt.scatter(x_disc, y_disc, marker='.', color='green')
plt.scatter(x_R, y_R, marker='.', color='b')
plt.legend(['Disk points (D)', 'Random points (R)'], loc='best')
plt.xlim(-500, 500)
plt.ylim(-500, 500)
plt.savefig('points_disk.jpg', dpi=300)
plt.show()
```

```python
#Realizando la lista de distancias entre los datos D y R
counter = 0
Dij_disc_l = []
Rij_disc_l = []
DRij_disc_l = []

for i,j,k,l in zip(x_disc, y_disc, x_R, y_R):
    counter += 1
    if len(x_disc) > counter:
        for m,n,p,q in zip(x_disc[counter:], y_disc[counter:], x_R[counter:], y_R[counter:]):
            D_ij = np.sqrt((i - m)**2 + (j - n)**2)
            R_ij = np.sqrt((k - p)**2 + (l - q)**2)
            Dij_disc_l.append(D_ij)
            Rij_disc_l.append(R_ij)

for i,j in zip(x_disc, y_disc):
    for k,l in zip(x_R, y_R):
        DR_ij = np.sqrt((i - k)**2 + (j - l)**2)
        DRij_disc_l.append(DR_ij)

Dij_disc = np.array(Dij_disc_l)
Rij_disc = np.array(Rij_disc_l)
DRij_disc = np.array(DRij_disc_l)
```

```python
#Escogiendo las distancias mínima y máxima para realizar el bineo de los histogramas para datos: DD y RR
Dij_min = np.min(Dij_disc)
Dij_max = np.max(Dij_disc)
Rij_min = np.min(Rij_disc)
Rij_max = np.max(Rij_disc)
if Dij_min > Rij_min:
    distance_min_disc = Rij_min
else:
    distance_min_disc = Dij_min
    
if Dij_max > Rij_max:
    distance_max_disc = Dij_max
else:
    distance_max_disc = Rij_max
```

```python
# Estableciendo los bines para los histogramas de los datos DD y RR
bines_disc = np.linspace(distance_min_disc, distance_max_disc, 34)
bines2_disc = np.linspace(distance_min_disc, distance_max_disc, 60)
```

```python
#Escogiendo las distancias mínima y máxima para realizar el bineo de los histogramas para datos: DD, DR, RR
DRij_min = np.min(DRij_disc)
DRij_max = np.max(DRij_disc)

if Dij_min > DRij_min:
    if DRij_min > Rij_min:
        distance_min_disc2 = Rij_min
    else:
        distance_min_disc2 = DRij_min
if DRij_min > Dij_min:
    if Dij_min > Rij_min:
        distance_min_disc2 = Rij_min
    else:
        distance_min_disc2 = Dij_min
    
if Dij_max > DRij_max:
    if Dij_max > Rij_max:
        distance_max_disc2 = Dij_max
    else:
        distance_max_disc2 = Rij_max

if DRij_max > Dij_max:
    if DRij_max > Rij_max:
        distance_max_disc2 = DRij_max
    else:
        distance_max_disc2 = Rij_max
```

```python
# Estableciendo los bines para los histogramas de los datos DD, DR y RR
bines_disc_2 = np.linspace(distance_min_disc2, distance_max_disc2, 34)
bines2_disc_2 = np.linspace(distance_min_disc2, distance_max_disc2, 60)
```

```python
# Obtención de los puntos DD y RR para calcular la función de correlación de dos puntos.
counts_DD_disc, bins = np.histogram(Dij_disc, bins = bines_disc, density=False) #normalized, son las coordenadas y 
counts_RR_disc, bins = np.histogram(Rij_disc, bins = bines_disc, density=False)
binscenters_disc = np.array([0.5 * (bines_disc[i] + bines_disc[i+1]) for i in range(len(bines_disc)-1)]) #coordinate of bin center son las coordenadas x del histograma
# Estimador de Peebles-Hauser para la fucnión de correlación de dos puntos
two_point_PH_disc = (counts_DD_disc/counts_RR_disc)-1
```
